# **YOLO Model Comparison**

# **1. Importing Dependencies**

```python
import pandas as pd
import matplotlib.pyplot as plt
```

# **2. CSV File Loading**

```python
df5  = pd.read_csv("/kaggle/input/model-1/5_validation_results.csv", index_col=0)
df10 = pd.read_csv("/kaggle/input/model-1/10_validation_results.csv", index_col=0)
df15 = pd.read_csv("/kaggle/input/model-1/15_validation_results.csv", index_col=0)
df20 = pd.read_csv("/kaggle/input/model-1/20_validation_results.csv", index_col=0)
```

# **3. Metrics Comparison**

```python
import pandas as pd
import matplotlib.pyplot as plt

metrics = ["Box-P","Box-R","Box-F1","mAP50", "mAP50-95"]

# Compute mean metrics
summary_df = pd.DataFrame({
    "5 Epochs":  [df5[m].mean()  for m in metrics],
    "10 Epochs": [df10[m].mean() for m in metrics],
    "15 Epochs": [df15[m].mean() for m in metrics],
    "20 Epochs": [df20[m].mean() for m in metrics],
}, index=metrics)

print(summary_df)

# Plot
summary_df.plot(kind="bar", figsize=(10, 6))

plt.title("YOLO Model Comparison Across Epochs")
plt.xlabel("Training Epochs")
plt.ylabel("Score")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.legend(title="Metric")
plt.tight_layout()
plt.show()
```

# **4. Classwise Comparsion**

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Class appending & serialization 
classes = df5.index.tolist()
x = np.arange(len(classes))
w = 0.18

plt.figure(figsize=(20, 20))

# Subplot 1: Box-P
plt.subplot(5, 1, 1)
plt.bar(x + 0*w, df5["Box-P"],  w, label="5 Epochs")
plt.bar(x + 1*w, df10["Box-P"], w, label="10 Epochs")
plt.bar(x + 2*w, df15["Box-P"], w, label="15 Epochs")
plt.bar(x + 3*w, df20["Box-P"], w, label="20 Epochs")
plt.title("Class-wise Precision")
plt.ylabel("Precision")
plt.legend(ncol=4)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.xticks([])

# Subplot 2: Box-R
plt.subplot(5, 1, 2)
plt.bar(x + 0*w, df5["Box-R"],  w)
plt.bar(x + 1*w, df10["Box-R"], w)
plt.bar(x + 2*w, df15["Box-R"], w)
plt.bar(x + 3*w, df20["Box-R"], w)
plt.title("Class-wise Recall")
plt.ylabel("Recall")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.xticks([])

# Subplot 3: Box-F1
plt.subplot(5, 1, 3)
plt.bar(x + 0*w, df5["Box-F1"],  w)
plt.bar(x + 1*w, df10["Box-F1"], w)
plt.bar(x + 2*w, df15["Box-F1"], w)
plt.bar(x + 3*w, df20["Box-F1"], w)
plt.title("Class-wise F1 Score")
plt.ylabel("F1 Score")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.xticks([])

# Subplot 4: mAP50
plt.subplot(5, 1, 4)
plt.bar(x + 0*w, df5["mAP50"],  w)
plt.bar(x + 1*w, df10["mAP50"], w)
plt.bar(x + 2*w, df15["mAP50"], w)
plt.bar(x + 3*w, df20["mAP50"], w)
plt.title("Class-wise mAP50")
plt.ylabel("mAP50")
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.xticks([])

# Subplot 5: mAP50-95
plt.subplot(5, 1, 5)
plt.bar(x + 0*w, df5["mAP50-95"],  w)
plt.bar(x + 1*w, df10["mAP50-95"], w)
plt.bar(x + 2*w, df15["mAP50-95"], w)
plt.bar(x + 3*w, df20["mAP50-95"], w)
plt.title("Class-wise mAP50-95")
plt.ylabel("mAP50-95")
plt.xticks(x + 1.5*w, classes, rotation=45, ha="right")
plt.xlabel("Classes")
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.show()
```
